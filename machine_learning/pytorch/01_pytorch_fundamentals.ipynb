# PyTorch Fundamentals

Pytorch has mainly five organs:

1. Tensor: The basic data structure in PyTorch, similar to NumPy arrays but with additional capabilities for GPU acceleration.

2. Autograd: PyTorch's automatic differentiation engine that allows for easy computation of gradients, which is essential for training neural networks.

3. nn: A high-level module that provides a wide range of pre-built layers, loss functions, and optimizers for building and training neural networks.

4. Loss Function: A function that measures the difference between the predicted output of a model and the actual target values, guiding the optimization process during training.

5. Optimizer: A module that implements various optimization algorithms (like SGD, Adam, etc.) to update the model's parameters based on the computed gradients during training.

Overall pipeline:

    1. Load data
    2. Send batch to model
    3. Model makes predictions
    4. Compare predictions with true labels
    5. Compute loss
    6. Backpropagate loss
    7. Optimizer updates weights
    8. Repeat many times
    9. Test on unseen data
    10. Save trained model



## 1. Tensors

A Tensor is the fundamental data structure in PyTorch. It represents numerical data, shape, memory layout, device, and gradient tracking.

```python
import torch

# Tensor: numerical data + shape + memory layout + device + gradient tracking

scalar = torch.tensor(1)

print(f"Scalar: {scalar}")
```

### Tensor Properties
You can inspect a tensor's shape, data type, and the device it is stored on (CPU or GPU).

```python
import torch

x = torch.tensor([1.0, 2.0, 3.0])

print(x)
print(x.shape)
print(x.dtype)
print(x.device)

"""
Output:
    tensor([1., 2., 3.])
    torch.Size([3])
    torch.float32
    cpu
"""


import torch

a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

print(a + b)

# output: tensor([5, 7, 9])
```

### Tensor Arithmetic
Tensors support standard arithmetic operations natively, usually performing them element-wise.

```python
import torch

data = [1, 2, 3]

print(torch.tensor(data))
```
