# Advanced String Formatting


Python offers incredibly powerful string interpolation tools. Here is a breakdown of all the ways you can format text, align outputs, and convert numerical representations directly inside f-strings.

### 00_str_format.py

```python
# 1. multiplication method
print(f"{'=' * 12}Your message{'=' * 12}")

# Output: ============Your Message============


# 2. center(width, fillchar): more readable approach
print("Your Message".center(36, "="))  # Output: ============Your Message============
print("Your Message".center(36, "-"))  # Output: ------------Your Message------------
print("Your Message".center(36, "*"))  # Output: ************Your Message************
print("Your Message".center(36, "#"))  # Output: ############Your Message############


# 3. Working with variables and spaces
index = 0
label = "Cat"
pred = "Cat"
conf = 0.8

print(f"Image {index + 1}: True = {label:12s} | Predicted = {pred:12s} | Confidence = {conf:.4f}")

print(f"Image {index + 1}: True = {label:12s} | Predicted = {pred:12s} | Confidence = {conf:.4f}")

"""
Image 1: True = Cat          | Predicted = Cat          | Confidence = 0.8000
Image 1: True = Cat          | Predicted = Cat          | Confidence = 0.8000
"""
```

### 01_str_alignment.py

```python
name = "Ali"

print(f"{name:10}")  # Left-aligned in width 10
print(f"{name:<10}")  # Explicit left align
print(f"{name:>10}")  # Right align
print(f"{name:^10}")  # Center align

"""
Ali       
Ali       
       Ali
   Ali
"""

print(f"{name:*<10}")  # Ali*******
print(f"{name:*>10}")  # *******Ali
print(f"{name:=^10}")  # ===Ali====
```

### 02_integers.py

```python
n = 42

# n:5 define no. of spaces
print(f"{n:5}")  # Output:    42
print(f"{n:05}")  # Output: 00042    zero-padded
print(f"{n:+5}")  # Output:   +42    Always show plus sign
print(f"{n:+5}")  # Output:   +42    Always show plus sign

n = 1000000
print(f"{n:,}")  # Output: 1,000,000     thousands separator
print(f"{n:_}")  # Output: 1_000_000     thousands separator

print(f"{-42:+}")  # Output: -42
```

### 03_floating.py

```python
e = 2.71828

print(f"{e:.2f}")  # Output: 2.72          -- two decimal places
print(f"{e:.4f}")  # Output: 2.7183        -- four decimal places
print(f"{e:10.2f}")  # Output:       2.72        -- width 10, 2 decimals
```

### 04_percentage.py

```python
rate = 0.853

print(f"{rate:.1%}")  # output: 85.3%
```

### 05_bin_oct_hex.py

```python
n = 178910

print(f"{n:b}")  # Output: 101011101011011110        -- binary
print(f"{n:o}")  # Output: 535336                    -- octal
print(f"{n:x}")  # Output: 2bade                     -- hexa lowercase
print(f"{n:X}")  # Output: 2BADE                     -- hexa Uppercase
```

### 06_sci_notation.py

```python
# Scientific notations

n = 12345678

print(f"{n:e}")  # 1.234568e+07              1.234568 × 10^7
print(f"{n:.2e}")  # 1.23e+07                 1.23 x 10^7


# common ML examples
lr = 1e-3
print(f"lr = {lr:.0e}")  # lr = 1e-03        1 x 10^-3

loss = 2.34567e-7
print(f"loss = {loss:.3e}")  # loss = 2.346e-07  2.346 x 10^-7


# Global formatter
"""
g automatically chooses between:

fixed point (f)
scientific notation (e)

depending on which looks better.
"""


x = 0.00001234
print(f"{x:g}")  # 1.234e-05


x = 1234.567
print(f"{x:g}")  # 1234.57

# Significant Digits
x = 0.000012345678

print(f"{x:.3g}")  # 1.23e-05              # Show 3 or 5 significant digits, not decimal places.
print(f"{x:.5g}")  # 1.234e-05
```

### 07_debug_fmt.py

```python
# Debugging formatting

name = "Bedar"
age = 21

print(f"{name=} | {age=}")  # Output: name='Bedar' | age=21
```

### 08_integer_d.py

```python
# d means decimal integer. Display this value as a base-10 integer.

n = 42

print(f"{n:d}")  # Output: 42        -- Usually you don't need d by itself because Python already knows it's an integer.


n = 7
f"{n:03d}"  # Output: 007
"""
0  -> pad with zeros
3  -> total width = 3
d  -> decimal integer
"""
```

### 09_date_fmt.py

```python
from datetime import datetime

now = datetime.now()

print(f"{now:%Y-%m-%d}")  # Output: 2026-06-25
print(f"{now:%H-%M-%S}")  # Output: 21-52-01
```

### 10_table.py

```python
name = "Ali"
score = 95

print(f"{'Name':<10}{'Score':>10}")
print(f"{name:<10}{score:>10}")

"""
Name           Score
Ali               95
"""
```
