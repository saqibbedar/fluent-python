## The `enumerate` Function

### 00_intro.py

```python
"""
Def: enumerate() is one of Python's most useful built-in functions. It lets you loop over an iterable while automatically keeping track of an index.

- Function signature: enumerate(iterable, start=0)

- Parameters:

    1. iterate (required):

        Any object that can be iterated over:

        list
        tuple
        string
        set
        dict
        generator
        file object
        range

        Examples:

            enumerate([10, 20, 30])
            enumerate("hello")
            enumerate(range(5))

    2. start (optional):

        default -> start=0

        Example:

            for i, item in enumerate(["a", "b", "c"], start=1):
                print(i, item)

            Output:

                1 a
                2 b
                3 c

- Negative values are allowed

    for i, item in enumerate(["a", "b", "c"], start=-3):
        print(i, item)

    Output:

        -3 a
        -2 b
        -1 c

- Return value: It returns an enumerate object.

    nums = [10, 20, 30]

    e = enumerate(nums)

    print(e)

    Output: <enumerate object at 0x...>

    Note: It does not immediately create all index-value pairs. It is a lazy iterator.

- Internally implemented like this:

    def custom_enumerate(iterable, start=0)
        n = start

        for item in iterable:
            yield (n, item)
            n += 1

    list(custom_enumerate(["a", "b", "c"]))

    output: [(0, 'a'), (1, 'b'), (2, 'c')]      # it return array of tuple so

    for tup in enumerate(iterable):
        print(tup)

    Output: (index, value)
"""
```

### 01_enum_with_strings.py

```python
# Usage with strings

for i, ch in enumerate("Bedar"):
    print(f"{i=}, {ch=}")

"""
i=0, ch='B'
i=1, ch='e'
i=2, ch='d'
i=3, ch='a'
i=4, ch='r'
"""
```

### 02_enum_with_tuples.py

```python
# Using with tuples

for i, value in enumerate((10, 20, 30)):
    print(i, value)

# Output:
# 0 10
# 1 20
# 2 30
```

### 03_enum_with_dic.py

```python
data = {"a": 100, "b": 200, "c": 300}

for index, key in enumerate(data):
    print(index, key)

# Output:
# 0 a
# 1 b
# 2 c

print(f"{'Index':<10} | {'Key':<10} | {'Value'}")
for i, (k, v) in enumerate(data.items()):
    print(f"{i:<10} | {k:<10} | {v}")

"""
Index      | Key        | Value
0          | a          | 100
1          | b          | 200
2          | c          | 300
"""
```

### 04_enum_with_file.py

```python
with open("00_enumerate.py") as f:
    if not f:
        print("Unable to read file")
    else:
        print(f"{'Line No.':<10} | Contents")
        for line_no, line_contents in enumerate(f, start=1):
            print(f"{line_no:<10} | {line_contents}")
```

### 04_enum_with_list.py

```python
# Note: You can start with any index, either negative or positive. If positive, indexes will grow orderly and decrease same.

# Positive: starting index is 5.
for i, val in enumerate(["apple", "banana", "carrot"], start=5):
    print(i, val)

"""
5 apple
6 banana
7 carrot
"""


# Positive: starting index is 5.
for i, val in enumerate(["apple", "banana", "carrot"], start=-5):
    print(i, val)

"""
-5 apple
-4 banana
-3 carrot
"""
```

### 05_enum_iterator.py

```python
# Enumerate object is an iterator

e = enumerate(["a", "b", "c"])

print(next(e))
print(next(e))
print(next(e))

"""
Output:
(0, 'a')
(1, 'b')
(2, 'c')
"""

# print(next(e))        # Error: raises StopIteration
```

### 06_consumed_iterator_behavior.py

```python
# Consumed Iterator Behavior

e = enumerate(["a", "b", "c"])


print(list(e))  # output: [(0, 'a'), (1, 'b'), (2, 'c')]
print(list(e))  # []: Because the iterator is already exhausted.
```

### 07_enum_with_generators.py

```python
# Enumerator with Generators: in python generators are special type of iterators that produce one value at a time, on demand, instead of generating and storing all values in memory at once.

# They are useful when working with large datasets, streams of data, or sequences that could be infinite.


def square():
    for i in range(5):
        yield i * i


for idx, value in enumerate(square()):
    print(idx, value)

"""
0 0
1 1
2 4
3 9
4 16
"""

# No extra memory is used to store all values.
```

## Itertools Module

In Python, `itertools.chain()` is a function that combines multiple iterables into a single iterator. It processes data sequentially without copying or duplicating the underlying elements in memory.

Instead of merging lists together and allocating brand-new memory, itertools.chain() lazily steps through each container one by one until they are all exhausted.

💡 Why Use itertools.chain?

- **Zero Memory Overhead**: It does not create a new combined list.

- **Handles Large Data**: It works flawlessly on massive datasets or endless streams.

- **Mix and Match**: It can seamlessly chain different data types together (e.g., a list, a tuple, and a set).

```python
from itertools import chain

# different data collections
fruits = ["apple", "banana"]
veggies = ("carrot", "potato")

# combine them smoothly
combined = chain(fruits, veggies)

for item in combined:
    print(item)

print(list(chain(fruits, veggies)))
```

---

# ⚡ The Shortcut: chain.from_iterable()

If your iterables are already packed inside a single nested list or container, unpacking them with chain(*list_of_lists) can slow your code down. Instead, use chain.from_iterable() to flatten them instantly and efficiently.

```python
from itertools import chain

nested_matrix = [[1, 2], [3, 4], [5, 6]]

flattened = list(chain.from_iterable(nested_matrix))

print(flattened)
```

> list(chain.from_iterable()) faster than list comprehension [x for row in matrix for x in row]

# 1. All Full-Length Permutations

By default, the function generates permutations matching the length of the input data.

```python
from itertools import permutations

# Elements can be list, tuple, or string
items = [1, 2, 3]
all_perms = list(permutations(items))
print(all_perms)
```

---

# 2. Permutations of a Specific Length

Pass an optional second argument `r` to specify the exact length of the arrangements.

```python
from itertools import permutations

short_perms = list(permutations(["a", "b", "c"], 2))

print(short_perms)
```

---

# 3. Permutations of a String

Because `permutations()` treats string inputs as a sequence of characters, you can join the resulting tuples back into strings using `.join()`.

```python
from itertools import permutations

word_perms = ["".join(p) for p in permutations("xyz")]

print(word_perms)
```

---

# 4. Permutations with Repetition (Cartesian Product)

If you want to allow items to repeat across positions (like picking combinations for a lock combination), use `itertools.product()`.

```python
from itertools import product

# Allows repeating elements like (1, 1)
repeat_perms = list(product([1, 2, 3], repeat=2))

print(repeat_perms)
```

---

# 5. Calculate Only the Total Number of Permutations

If you only need to know how many permutations exist without listing them out, use `math.perm(n, k)`.

```python
import math

# Total ways to arrange 3 items out of a pool of 5
total_ways = math.perm(5, 3)

print(total_ways)
```
