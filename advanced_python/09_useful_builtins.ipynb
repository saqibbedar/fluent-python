# The `timeit` Module

The `timeit` module is Python's built-in utility designed specifically to measure the execution time of small code snippets with extreme accuracy. It prevents common benchmarking traps by automatically selecting the most precise system clock and temporarily disabling background garbage collection.

### Core Features:

- **Avoids OS Noise**: It executes code thousands or millions of times to average out background system spikes.

- **Isolates Setup**: It separates initialization logic (like imports) from the code being actively measured.

- **Multiple Interfaces**: It can be used directly inside Python scripts, via the command line, or inside Jupyter Notebooks.

---

### Option A: Passing Code as Strings (Traditional)

````python
import timeit

# Code to test is passed to 'stmt'
# Setup operations (like imports) are passed to 'setup'
execution_time = timeit.timeit(stmt="[x**2 for x in range(100)]", setup="pass", number=100000)

print(f"Total time for 100,000 runs: {execution_time:.4f} seconds")
````

### Option B: Passing Functions (Recommended for Readability)

````python
import timeit


def test_logic():
    return list(map(lambda x: x**2, range(100)))


# Pass the function directly. 'number' defaults to 1,000,000 if omitted.
execution_time = timeit.timeit(test_logic, number=100000)
print(f"Total time: {execution_time:.4f} seconds")
````

### Option C: Accessing Local Variables

````python
import timeit

data_list = list(range(1000))

# 'globals=globals()' gives the snippet access to 'data_list'
run_time = timeit.timeit(stmt="len(data_list)", globals=globals(), number=1000000)
print(f"Execution time: {run_time:.4f} seconds")
````

# Using timeit via Command Line (CLI)

```bash
# General syntax
python -m timeit "[x**2 for x in range(100)]"

# Using a setup statement (-s) to handle imports
python -m timeit -s "import math" "math.sqrt(144)"
```

# 1. Universally Unique Identifiers (UUIDv4)

It generates a standard 128-bit (36-character) string. It takes virtually zero memory because it uses optimized C-extensions under the hood.

````python
import uuid

# Efficient string format
unique_id = str(uuid.uuid4())
print(unique_id)  # Output: '46cae811-1cfc-487a-b8b7-3969fb9c281b'
````

````python
import secrets
import string


def generate_short_id(length=12):
    # Alphabet: Letters (upper/lower) + Digits (62 total characters)
    alphabet = string.ascii_letters + string.digits

    # Securely choose characters and join them
    return "".join(secrets.choice(alphabet) for _ in range(length))


# Example Usage
print(generate_short_id(12))  # Output: 'J67pSyhIrKgw'
````

# Platform info

Python has many implementation versions, the core ships by python.org is CPython.

Other distributions:

    - PyPy: written in RPython, with a JIT compiler. Instead of interpreting bytecode forever, it watches your hot loops and compiles them to actual machine code at runtime.

    - Jython: Python on JVM, compiles to Java bytecode, can use java libraries. Largely stuck on Python 2-era compatibility; mostly of historical interest now.

    - IronPython: same idea for .NET/C#.

    - MicroPython / CircuitPython: a lean reimplementation for microcontrollers with kilobytes of RAM.

    - GraalPy: Oracle's implementation on GraalVM, modern JIT approach, improving C-extension support.

    - RustPython: a community implementation in Rust.
