# `shutil` The Ultimate guide

The Python shutil (shell utilities) module provides high-level operations for copying, moving, renaming, and deleting files and directories. While the os module handles low-level system interactions, shutil automates complex tasks like recursive directory copying or archiving with a single command.

### Quick Reference Table

| Function | Action | Best Used For |
|---|---|---|
| `shutil.copy(src, dst)` | Copies file content and permissions | Basic file copying |
| `shutil.copy2(src, dst)` | Copies file content and metadata (timestamps) | Creating backups while preserving dates |
| `shutil.copytree(src, dst)` | Recursively copies an entire directory | Backing up folders |
| `shutil.move(src, dst)` | Moves or renames files and folders | Organizing files into new directories |
| `shutil.rmtree(path)` | Recursively deletes a directory tree | Deleting folders containing files |
| `shutil.disk_usage(path)` | Returns total, used, and free storage space | Pre-checking storage limits |

---

# Code Examples

**1. Copying Files and Folders**

```python
import shutil

# Copy a single file (preserves contents and permissions)
shutil.copy("./files/file.txt", "./files/file2.txt")
```

```python
# Copy a file while keeping creation/modification timestamps
shutil.copy2("./files/file2.txt", "./files/backup.txt")
```

```python
# Copy an entire directory tree to a new location
shutil.copytree("./files", "./backup_files")
```

---

**2. Moving and Deleting**

```python
# Move a file or folder (acts as a rename if target path is in the same folder)
shutil.move("./files/data.csv", "backup_files/data.csv")
```

```python
# Force delete a directory even if it contains files (Warning: Permanent!)
shutil.rmtree("backup_files")
```

---

**3. Creating and Extracting Archives (ZIP / TAR)**

```python
# Compress an entire directory into a single zip file
shutil.make_archive("backup_archive", "zip", "./files")
```

```python
# Extract the archive back into a folder
shutil.unpack_archive("backup_archive.zip", "extracted_files")
```

---

**4. Checking Disk Space**

```python
# Check remaining storage space on a specific drive
total, used, free = shutil.disk_usage("/")
print(f"Free space: {free / (1024**3):.2f} GB")
```
