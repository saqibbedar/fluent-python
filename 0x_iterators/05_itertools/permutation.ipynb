# 1. All Full-Length Permutations

By default, the function generates permutations matching the length of the input data.

```python
from itertools import permutations

# Elements can be list, tuple, or string
items = [1, 2, 3]
all_perms = list(permutations(items))
print(all_perms)
```

---

# 2. Permutations of a Specific Length

Pass an optional second argument `r` to specify the exact length of the arrangements.

```python
from itertools import permutations

short_perms = list(permutations(['a', 'b', 'c'], 2))

print(short_perms)
```

---

# 3. Permutations of a String

Because `permutations()` treats string inputs as a sequence of characters, you can join the resulting tuples back into strings using `.join()`.

```python
from itertools import permutations

word_perms = [''.join(p) for p in permutations('xyz')]

print(word_perms)
```

---

# 4. Permutations with Repetition (Cartesian Product)

If you want to allow items to repeat across positions (like picking combinations for a lock combination), use `itertools.product()`.

```python
from itertools import product

# Allows repeating elements like (1, 1)
repeat_perms = list(product([1, 2, 3], repeat=2))

print(repeat_perms)
```

---

# 5. Calculate Only the Total Number of Permutations

If you only need to know how many permutations exist without listing them out, use `math.perm(n, k)`.

```python
import math

# Total ways to arrange 3 items out of a pool of 5
total_ways = math.perm(5, 3)

print(total_ways)
```
