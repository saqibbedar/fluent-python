# 1. Initialize the Generator

Start by initializing the generator object. You can optionally pass a specific integer value as a seed to ensure your random outputs are exactly the same every time you run the script.

```python
import numpy as np

# For production/varying outputs:
rng = np.random.default_rng()

# For reproducible results (testing/sharing)
rng = np.random.default_rng(seed=42)
```

---

# 2. Common Random Number Methods

Once you have your rng object, use these core methods to generate data or sample from arrays:

- **`rng.random(size)`**: Generates floating-point numbers uniformly distributed between `0.0` (inclusive) and `1.0` (exclusive) `[0.0, 1.0)`.

- **`rng.integers(low, high, size)`**: Generates random integers from `low` (inclusive) up to `high` (exclusive).

- **`rng.uniform(low, high, size)`**: Generates continuous floating-point numbers within a specific customized range.

- **`rng.normal(loc, scale, size)`**: Draws random samples from a normal (Gaussian) distribution, where `loc` is the mean and `scale` is the standard deviation.

- **`rng.choice(array, size, replace)`**: Randomly selects items from an existing array. Set replace=False to sample without duplicates.

- **`rng.shuffle(array)`**: Randomly shuffles the elements of an existing multi-dimensional array in-place.

# 3. Practical Code Examples

```python
# Generate a single random float between 0 and 1
single_float = rng.random()

# Generate a 1D array of 5 random floats between 0 and 1
float_array = rng.random(size=5)

# Generate a 2D array (shape 3x4) of random integers between 10 and 99
int_matrix = rng.integers(low=10, high=100, size=(3, 4))

# Generate 1000 numbers following a standard normal distribution (mean=0, std=1)
normal_dist = rng.normal(loc=0, scale=1, size=1000)

# Randomly select 3 unique elements from a list of names
participants = np.array(["Alice", "Bob", "Charlie", "David", "Eve"])
winners = rng.choice(participants, size=3, replace=False)
```

```python
rng = np.random.default_rng()

arr = np.full(shape=10, fill_value=2.3)

# Generates 10 distinct random numbers and multiplies element-by-element
arr = rng.random(size=arr.size)

print(arr)
```
