# namedtuple

A `namedtuple` is a factory function from Python's collections module that creates **immutable, lightweight, tuple-like objects** with named fields. This allows you to access elements using readable dot notation (e.g., `car.color`) instead of strictly relying on numeric indices (e.g., `car[0]`).

It offers the memory efficiency of a standard tuple while making your code significantly cleaner and self-documenting

---

### Basic Syntax & Creation

To use `namedtuple`, you first define the template structure and then instantiate objects from it.

```py
# first import it from collections 
from collections import namedtuple

# define template/structure
namedtuple(typename, field_names, *, rename=False, defaults=None, module=None)
```

**Argument Breakdown**:

- `typename`: The string name of the new tuple subclass.

- `field_names`: A list of strings `['x', 'y']`, a space-separated string `'x y'`, or a comma-separated string `'x, y'`.

- `rename`: If `True`, invalid or duplicate field names are automatically replaced with positional names like `_1`.

- `defaults`: An iterable of default values applied to the rightmost fields.

---

### Examples

**1. Basic Usage (Three different ways to define fields)**

````python
from collections import namedtuple

# Standard syntax using a list of strings
Point = namedtuple("Point", ['x', 'y'])

# Alternative: Space-separated string
Point = namedtuple("Point", "x y")

# Alternative: Comma-separated string
Point = namedtuple('Point', 'x, y')

# Creating an instance
p = Point(10, 20)

# Accessing values
print(p.x)     # Output: 10 (Using name)
print(p[1])    # Output: 20 (Using index)
````

**2. Using Default Values**

````python
from collections import namedtuple

# Defaults are applied from right to left (y gets 0, x gets 1)
Point = namedtuple('Point', ['x', 'y'], defaults=[1, 0])

p1 = Point()          # Point(x=1, y=0)
p2 = Point(5)         # Point(x=5, y=0)
p3 = Point(x=7, y=8)  # Point(x=7, y=8)
````

### Modern Alternative: Class-Based Syntax

For modern Python code bases requiring type hints, use `typing.NamedTuple` instead:

````python
from typing import NamedTuple

class Point(NamedTuple):
    x: int
    y: int

    units: str = "meters"   # Supports defaults and type hints
````

### More Examples

````python
from collections import namedtuple

# Define the structure and named attributes
Person = namedtuple('Person', ['name', 'age', 'height', 'city', 'country'])

# Create a named tuple from above defined namedtuple datastructure
p1 = Person("Saqib Bedar", 22, 5.4, "Islamabad", "Pakistan")

# Access via attribute name
print(p1.name)
print(p1.age)
````

**Updating attributes**

Python's `namedtuple` are immutable meaning they cannot be changed after creation. Hence, we cannot update attributes directly like `saqib.name = "New name"`.

If you try to do that, Python will raise an `AttributeError`.

Then how we do it? 

**Use the ._replace() Method**:

This is the official and cleanest way. It does not modify the original object; instead, it creates a brand new tuple instance with the updated values.

````python
from collections import namedtuple

# define structure for person
Person = namedtuple('Person', ['name', 'age', 'height', 'city', 'country'])
p1 = Person("Saqib", 22, 5.4, "Islamabad", "Pakistan")

# Create a new instance with the updated name
p1 = p1._replace(name="Saqib Bedar")

print(p1.name)  # Output: New name
print(p1.city)  # Output: Islamabad (other fields remain unchanged)
````

### lternative: Use `class` or `dataclasses` (If you need mutability)

````python
class Person:
    def __init__(self, name, age, height, city, country) -> None:
        self.name = name
        self.age = age
        self.height = height
        self.city = city
        self.country = country

# Or using dataclass with less boilerplate code

from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    height: float
    city: str
    country: str
````

````python
# Create the object

p1 = Person("Saqib", 22, 5.4, "Islamabad", "Pakistan")

# Modify the attributes directly
p1.name = "Saqib Bedar"
p1.city = "Larkana"
````
