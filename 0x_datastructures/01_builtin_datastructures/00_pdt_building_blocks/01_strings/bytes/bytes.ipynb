# bytes

An immutable sequence of integers ranging from 0 to 255. Primarily used to handle machine-readable binary data—such as image files, network packets, or text that has been encoded into a specific format (like UTF-8 or ASCII).

---

# Creating a bytes Object

You can initialize a bytes object in three common ways:

1. **Using a literal prefix (b'')**: Place a b directly before a string literal.

2. **Encoding a standard string**: Convert a readable string into a byte sequence.

3. **Using the bytes() constructor**: Pass an integer (for a sequence of null bytes) or an iterable of numbers.

```python
# 1. Byte literal syntax
b1: bytes = b"Hello"

print(b1)
print(type(b1))
```

```python
# 2. Encoding a string
b2: bytes = "Hello".encode("utf-8")

print(b2)
```

```python
# 3. Constructor with an iterable of integers (0-255)
b3: bytes = bytes([72, 101, 108, 108, 111])  # ASCII numbers for 'Hello'

print(b3)
```

```python
# 4. Constructor with a size (creates empty null bytes)
b4: bytes = bytes(5)

print(b4)
```

# Key Behaviors and Pitfalls

### 1. Elements are Integers, Slices are Bytes

When you index a single item in a bytes object, Python returns its underlying integer representation (0–255). However, slicing returns a new bytes object.

```python
b_data: bytes = b"Python"

print(b_data[0])  # Output: 80  (The ASCII value for 'P')
print(b_data[:2])  # Output: b'Py' (A slice returns a bytes object)
```

# 2. Conversion (Encoding & Decoding)

You cannot implicitly mix text strings and byte objects (e.g., b"Hello" + "World" will raise a TypeError). You must explicitly convert between them:

```python
# str -> bytes
my_bytes: bytes = "Bedar".encode("utf-8")  # output: b'Bedar'
print(my_bytes)

# bytes -> Str
my_str: str = my_bytes.decode("utf-8")
print(my_str)  # Bedar
```

# 3. Hexadecimal Representation

When non-printable ASCII characters are contained inside a byte sequence, Python represents them using hex escape codes (\x..):

```python
# creating bytes directly from a hex string
b_hex: bytes = bytes.fromhex("48 65 6c 6c 6f")

print(b_hex)

# turning bytes back to hex string
print(b_hex.hex())
```
