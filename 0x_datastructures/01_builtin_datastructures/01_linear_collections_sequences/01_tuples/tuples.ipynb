# tuple

A built-in data collection type used to store multiple items in a single variable. The defining characteristic of a tuple is that it is ordered (maintains sequence) and immutable (cannot be changed, altered, or edited after creation).

- **Mutable**: The data inside the object can change, but its **memory address stays the same**. (Think of a dry-erase whiteboard—you can change the message, but the board itself remains in the same spot). 

- **Immutable**: The data cannot change. Any change creates a **brand-new object at a new memory address**. (Think of a stone tablet—you cannot edit the message; you must carve a completely new tablet).

---

### 1. Creating a Tuple

Tuples are created using round brackets `()` with items separated by commas.

```python
# Empty tuple
empty_tuple = ()

# Tuple with integers
numbers = (1, 2, 3, 4, 5)

# Mixed data types
user_profile = ("Saqib Bedar", 22, True)

# Creating a tuple without brackets (Tuple unpacking)
coordinates = 10.5, 20.3
```

⚠️ **The Single-Item Trap**: To create a tuple with only one item, you must include a trailing comma. Without it, Python treats it as a regular string or number wrapped in parentheses.

```python
not_a_tuple = ("Apple")         # type: str
is_a_tuple = ("Apple",)         # type: tuple

print(type(not_a_tuple))
print(type(is_a_tuple))
```

---

### 2. Accessing Elements

Tuples are zero-indexed, meaning the first item is at index `0`. You can access elements using square brackets `[]`.

```python
fruits_tuple = ("Apple", "Banana", "Carrot", "Date")

print(fruits_tuple[0])          # Apple
print(fruits_tuple[-1])         # Date  (Negative indexing counts from the back)
```

---

### 3. Slicing a tuple

You can extract a specific range of elements by defining a `[start:stop:stride/step]` range.

> Note: Indexing works same as to the lists. 

```python
print(fruits_tuple[1:3])
print(fruits_tuple[0::2])
print(fruits_tuple[-4:])
print(fruits_tuple[-4:-1])
print(fruits_tuple[::-1])
```

---

### 3. The Core Rule: Immutability

Once a tuple is initialized, you cannot add, remove, or change its values.

```python
modes = ("Light", "Dark")

try:
    modes[0] = "System"
except TypeError as te:
    print(f"Error: cannot assign a value at index 0.")
    print(te)
```

---

### 4. Tuple Unpacking

You can extract the values of a tuple directly into individual variables.

```python
point = (10, 20)

x, y = point        # Unpacking

print(x, y)

scores = (1, 2, 2, 2, 3, 4, 5, 5, 5, 6)


print("-"*10)

# *rest: captures as list
first, second, *rest = scores

print(first)
print(second)
print(rest)


print("-"*10)

# or get another way
first, *middle, last = scores

print(first)
print(middle)
print(last)
```

```python
# A tuple representing a matrix row: ID, Score1, ID, Score2, ID, Score3...
survey_data = ("UserA", 95, "UserB", 88, "UserC", 92, "UserD", 79)

# Slice for scores only (index 1, 3, 5, 7) and unpack them instantly
first_score, *other_scores, last_score = survey_data[1::2]

print(first_score)
print(other_scores)
print(last_score)
```

---

### 5. Built-in Tuple Methods

Because tuples cannot be altered, they only feature two built-in methods:

- `.count(value)`: Returns the number of times a value appears.

- `.index(value)`: Returns the index position of the first occurrence of a value.

```python
scores = (1, 2, 2, 2, 3, 4, 5, 5, 5, 6)

print(scores.count(2))              # frequency/appearance of 2 in scores tuple
print(scores.index(2))              # returns first matched item index
```

### 6. Looping over tuple (All methods)

**6.1: The Standard `for` Loop (Most Common)**

```python
fruits = ("apple", "banana", "cherry", "date")

for fruit in fruits:
    print(fruit)
```

**6.2: The enumerate() Loop (Items + Index)**

```python
brands = ("Apple", "Google", "Microsoft")

for index, brand in enumerate(brands):
    print(index, brand)
```

> Tip: You can change the starting index by adding a parameter: enumerate(brands, start=1).

**6.3: The range(len()) Loop (By Index Only)**

```python
colors = ("Red", "Green", "Blue")

for index in range(len(colors)):
    print(f"Item at position {index} is {colors[index]}")
```

**6.4: Looping Over Zipped Tuples (Parallel Iteration)**

```python
names = ("Saqib", "Ali", "Hamza", "Abdullah")
ages = (22, 23, 22, 22)

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
```

**6.5: One-Line Looping (Tuple Comprehension)**

```python
numbers = (1, 2, 3, 4)

# Square every number in the tuple on a single line
squared_numbers = tuple(num**2 for num in numbers)

print(squared_numbers)
```

### 7. Usage in `dict`

In Python, a dictionary key must be **hashable**, meaning its value cannot change over its lifetime. Because **tuples are immutable**, Python can generate a permanent unique "hash code" for them. Lists cannot do this because they can change at any time.

```python
map_locations = {
    (0, 0): "Starting Village",
    # Sindhi: village names
    (1, 5): "Sono Khan Chandio",
    (10, 2): "Bhanbho Khan Chandio",
}

# Looking up a location using a tuple
current_position = (1, 5)
print(map_locations[current_position])

# direct passing
print(map_locations[(10, 2)])
```

**The Sub-Item Trap**: A tuple is only hashable **if all of its internal elements are also immutable**. If you put a mutable object (like a list) inside a tuple, you can no longer use that tuple as a dictionary key.

```python
# ❌ This tuple contains a list, so it cannot be a dictionary key
invalid_key = (1, 2, [3, 4])

try:
    my_dict = {invalid_key: "Data"}
    print(my_dict[invalid_key])
except TypeError as te:
    print("Error: Unhashable type.")
    print(f"Error: {te}")
```
