Python uses **TimSort** on lists. t was invented by Tim Peters in 2002 specifically for Python, though it is so brilliant that it has since been adopted by Java, Android, and V8 (JavaScript).

Timsort is a **hybrid algorithm** that combines the best parts of **Merge Sort** and **Insertion Sort**.

---

**Time & Space Complexity**:

- Best case: *O(N)* when the list is already sorted.

- Average and Worst case: *O(N log N)*

- Space Complexity: *O(N)* auxiliary space.

- 

---

**In-Place Sorting**:

It means it mutates the existing list object. It does not create a brand new list.

Because `list.sort()` mutates the original list rather than creating a new one, it returns None to prevent you from accidentally overwriting your variable with a new assignment.

---

**The Arguments**:

The `list.sort()` method accepts two optional keyword-only arguments:

- **`key`**: A function that acts as a basis for sort comparison.

- **`reverse`**: A boolean value (True or False), default value is False.

---

### **Examples**

```python
# Sort by string length
words = ["banana", "apple", "cherry"]
words.sort(key=len)

print(words)
```

```python
# Sort a list of dictionaries by a specific key

data = [{"name": "Max", "age": 25}, {"name": "Alex", "age": 20}]

data.sort(key=lambda x: x["age"])

print(data)
```

```python
# The reverse Argument

numbers = [3, 1, 4, 2]
numbers.sort(reverse=True)

print(numbers)
```
