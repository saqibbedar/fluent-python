# Question 1

````python
m = [[0]*3]*3
m[1] = [7, 7, 7]
m[0][0] = 99
print(m)
````

**Explanation**: when `m` is created using trap syntax. The outerList's all three pointers will point to single list and it has `[0,0,0]`. So, there is only single inner list but three pointers from outerList are pointing at it. 

When we write `m[1] = [7, 7, 7]`, it will flip the pointer arrow from pointing at `[0, 0, 0]` to point at `[7, 7, 7]`. So, outerList has two inner list now. Where `ptr[1]` is pointing to list `[7,7,7]` while `ptr[0]` and `ptr[2]` still pointing at list `[0, 0, 0]`. 

And if we write `m[0][0] = 99`, we know our outer list has two list's and on index `[0]` we our `ptr[0]` is pointing to list `[0, 0, 0]` and `ptr[2]` is also pointing to that list. So, we assign `m[0][0] = 99` at the first index of list `[0, 0, 0]` and this list becomes `[99, 0, 0]`. 

💡 Worth to remember: Because our two pointers `ptr[1]` and `ptr[2]` were pointing to list `[0, 0, 0]` and change on this list will be reflected in both and because list is same and pointer arrows on it. 

Hence, output gonna be:

```py
[
    [99, 0, 0],
    [7, 7, 7],
    [99, 0, 0]
]
```

## Full Breakdown

When `m` is created using the trap syntax:

```python
m = [[0] * 3] * 3
```

the outer list contains **three references to the same inner-list object**:

```text
m[0] ──┐
m[1] ──┼──► [0, 0, 0]
m[2] ──┘
```

So, there is only **one inner list object**, but the outer list has three references pointing to it.

---

### `m[1] = [7, 7, 7]`

When we write:

```python
m[1] = [7, 7, 7]
```

we are **replacing the reference stored at index `1` of the outer list**.

The reference at `m[1]` no longer points to the original `[0, 0, 0]` list. It now points to a **new list**:

```text
m[0] ─────► [0, 0, 0]

m[1] ─────► [7, 7, 7]

m[2] ─────► [0, 0, 0]
```

Therefore, the outer list now references **two different inner-list objects**:

* `m[0]` and `m[2]` reference the same `[0, 0, 0]` object.
* `m[1]` references the new `[7, 7, 7]` object.

---

### `m[0][0] = 99`

Now we write:

```python
m[0][0] = 99
```

First, `m[0]` gives us the inner list:

```python
[0, 0, 0]
```

Then `[0]` accesses its first element:

```text
m[0]
  │
  ▼
[0, 0, 0]
 │
 └── [0] → 0
```

So assigning:

```python
m[0][0] = 99
```

changes that shared inner list from:

```python
[0, 0, 0]
```

to:

```python
[99, 0, 0]
```

But remember: **both `m[0]` and `m[2]` point to this same list object**.

Therefore, the change is visible through both references:

```text
m[0] ──┐
       ├──► [99, 0, 0]
m[2] ──┘

m[1] ─────► [7, 7, 7]
```

### Important Point

The original explanation might say that the "pointer arrow flips." A more precise way to say this is:

> **The reference stored at index `1` of the outer list is replaced with a reference to a new list.**

Also, the shared-list mutation happens because `m[0]` and `m[2]` still reference the **same mutable list object**.

## Final Output

Therefore:

```python
print(m)
```

produces:

```python
[
    [99, 0, 0],
    [7, 7, 7],
    [99, 0, 0]
]
```

Or, on one line:

```python
[[99, 0, 0], [7, 7, 7], [99, 0, 0]]
```

---

## Interview Takeaway

When solving nested-list questions, distinguish between these two operations:

### Replacing a reference

```python
m[1] = [7, 7, 7]
```

This replaces what the outer list's index `1` references.

### Mutating a shared object

```python
m[0][0] = 99
```

This modifies the inner list itself.

If another reference points to that same inner list, it will observe the modification too.

The key question to ask is:

> **Am I replacing a reference, or am I mutating an object that may be shared by multiple references?**
