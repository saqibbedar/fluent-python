```python
def f(x=[]):
    x.append(1)
    return x

print(f())
print(f())
print(f())
```

Answer: [] is created when the function definition is evaluated, not on every function call. Each call that omits x therefore receives a reference to the same list object. Since append() mutates that list, the mutation persists across calls.

---

```python
def f(x=[]):
    x.append([])
    return x

a = f()
b = f()

a[0].append(10)

print(a)
print(b)
print(a is b)
print(a[0] is b[0])
```
