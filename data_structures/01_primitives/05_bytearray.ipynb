# bytearray

The `bytearray` type is a mutable sequence of integers ranging from 0 to 255. It functions almost identically to [`bytes`](../bytes/bytes.ipynb), but with one major difference: you can modify its elements in place without creating a new object in memory.

---

# 1. Creating a `bytearray` Object

You can initialize a `bytearray` using the `bytearray()` constructor in several ways:

```python
# 1. From a string (must specify an encoding)
ba1: bytearray = bytearray("Hello", "utf-8")
print(ba1)  # Output: bytearray(b'Hello')

# 2. From an existing bytes object
ba2: bytearray = bytearray(b"World")
print(ba2)

# 3. From an iterable of integers (0-255)
ba3: bytearray = bytearray([72, 101, 108, 108, 111])  # ASCII for 'Hello'
print(ba3)

# 4. With a specific size (initializes with null bytes)
ba4: bytearray = bytearray(5)  # Output: bytearray(b'\x00\x00\x00\x00\x00')
print(ba4)
```

---

# 2. Modifying a `bytearray` (In-Place Mutability)

Because it is mutable, bytearray supports list-like operations such as item assignment, appending, popping, and extending.

```python
ba: bytearray = bytearray(b"Python")

# 1. Change an element using an integer (ASCII 74 = 'J')
ba[0] = 74
print(ba)  # Output: bytearray(b'Jython')

# 2. Append a single byte value
ba.append(33)  # ASCII 33 = '!'
print(ba)  # Output: bytearray(b'Jython!')

# 3. Extend with another byte sequence
ba.extend(b" Rules")
print(ba)  # Output: bytearray(b'Jython! Rules')

# 4. Delete elements using slicing
del ba[6:8]
print(ba)  # Output: bytearray(b'JythonRules')
```
