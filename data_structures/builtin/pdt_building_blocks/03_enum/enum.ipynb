# enum

An `Enum` (enumeration) in Python is a data type built into the standard enum module that helps you define a set of named symbolic constants. They make code clean, readable, and safe by replacing error-prone magic numbers or random strings.

---

# Creating and using basic Enums

```python
from enum import Enum


# Enum: standard baseclass
class Color(Enum):
    RED = 1
    GREEN = 2
    BLUE = 3


# accessing by name and value
print(Color.RED.name)
print(Color.RED.value)
print(Color.GREEN.name)
print(Color.GREEN.value)
print(Color.BLUE.name)
print(Color.BLUE.value)
```

---

# Iterating over Enum

```python
for clr in Color:
    print(f"{clr.name}: {clr.value}")
```

---

# Converting to a dict

```python
# dict comprehension
clr_enum_dict = {clr.name: clr.value for clr in Color}

print(clr_enum_dict)
```

# Converting to a list

```python
# save values
clr_values_list = [clr.value for clr in Color]
print(clr_values_list)
```

# Converting to a set

```python
unique_values = {clr.value for clr in Color}
print(unique_values)
```

# Common Types

## 1. IntEnum (Integer Enums)

An IntEnum is a subclass of both int and Enum. It is ideal when your enum values need to act exactly like numbers (e.g., database IDs, HTTP status codes, or protocol versions).

- **Key Feature**: Supports direct comparison with regular integers and basic math.

```python
from enum import IntEnum


class Priority(IntEnum):
    LOW = 1
    MEDIUM = 2
    HIGH = 3


# Direct comparison with regular integers
print(Priority.HIGH == 3)
print(Priority.LOW < 2)
```

## 2. StrEnum (String Enums)

Introduced in Python 3.11, StrEnum is a subclass of both str and Enum. It is perfect for configuration constants, file modes, or API payload strings.

- **Key Feature**: Behaves like a regular string. It serializes natively to JSON without needing custom encoders.

```python
from enum import StrEnum


class FileMode(StrEnum):
    READ = "r"
    WRITE = "w"
    APPEND = "a"


# Direct comparison with strings
print(FileMode.READ == "r")  # Output: True
print(f"Opening file with mode: {FileMode.WRITE}")  # Output: Opening file with mode: w
```

## 3. Flag and IntFlag (Bitwise Enums)

A Flag allows you to combine multiple enum values using bitwise operators (|, &). It is ideal for permissions, system settings, or multi-selection states.

- **Key Feature**: Members represent binary bits (1, 2, 4, 8, etc.). You can use auto() to set these automatically.

```python
from enum import Flag, auto


class Permission(Flag):
    READ = auto()  # 1
    WRITE = auto()  # 2
    EXECUTE = auto()  # 4


# Combining permissions using the OR (|) operator
user_perm = Permission.READ | Permission.WRITE

print(Permission.READ in user_perm)  # Output: True
print(Permission.EXECUTE in user_perm)  # Output: False
print(user_perm)  # Output: Permission.READ|WRITE
```

## 4. Useful Utilities: auto() and @unique

### Using auto()

If you do not care about the exact values, use auto() to assign values incrementally. For standard Enum, it assigns integers starting at 1. For StrEnum, it automatically assigns the lowercase string of the variable name.

```python
from enum import Enum, auto


class Status(Enum):
    PENDING = auto()  # 1
    ACTIVE = auto()  # 2
```

### Using @unique

By default, Enums allow different names to have the same value (creating an alias). If you want to strictly prevent this, use the @unique decorator.

```python
from enum import Enum, unique


@unique
class UniqueColor(Enum):
    RED = 1
    BLUE = 2
    # LIME = 1  <- This would trigger a ValueError because 1 is already taken
```
