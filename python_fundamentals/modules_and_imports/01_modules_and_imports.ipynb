# Modules and Imports

In Python, a **module** is simply a file containing Python code (like functions, classes, and variables). As your projects grow, you'll want to split your code across multiple files so it stays organized. 

Let's learn how to create our own modules and import them!

### 1. Creating a Module

Jupyter Notebooks have a special "magic command" called `%%writefile`. When you put this at the very top of a cell, it saves everything inside the cell directly to a file instead of executing it here.

Let's create a file named `math_utils.py`.

```python
%%writefile math_utils.py
def add(x: int, y: int):
    return x + y

def sub(x: int, y: int):
    return x - y

def mul(x: int, y: int):
    return x * y

def div(x: int, y: int):
    if y == 0:
        raise ZeroDivisionError("Cannot divide by zero!")
    else:
        return x / y
```

### 2. Importing the Module

Now that `math_utils.py` exists in our current directory, we can import it just like any built-in Python library.

> **Tip:** You can use the `as` keyword to give a module an alias. This is very common for saving keystrokes (e.g., `import numpy as np`).

```python
import math_utils as mu

x: int = 10
y: int = 0

print("Add:", mu.add(x, 5))
print("Sub:", mu.sub(x, 5))
print("Mul:", mu.mul(x, 5))
```

### 3. Handling Exceptions from Modules

If an error occurs inside an imported function, you can catch it in your main code using a `try/except` block.

```python
try:
    # This will trigger our custom ZeroDivisionError check inside math_utils.py
    print("Div:", mu.div(x, y))
except ZeroDivisionError as error:
    print(f"Caught an error: {error}")
finally:
    print("Execution completed.")
```

### 4. Cleanup
Since we only needed `math_utils.py` for this demonstration, let's clean it up using some Python standard libraries.

```python
import os
if os.path.exists("math_utils.py"):
    os.remove("math_utils.py")
    print("math_utils.py has been removed.")
```
