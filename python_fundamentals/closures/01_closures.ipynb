# Closures in Python

Welcome to the world of **closures**! This is one of Python's most powerful, yet often misunderstood, features.

## What is a Closure?
A closure is simply a nested function that remembers the variables from its enclosing (outer) scope, **even after the outer function has finished executing**.

Think of it like a backpack. The inner function packs up the variables it needs from the outer function, puts them in its backpack, and carries them around wherever it goes.

### 1. The Simplest Example
Let's see how an inner function captures the variable `x` from the outer function.

```python
def outer():
    x = 10

    def inner():
        return x  # 'x' is captured from the outer scope

    return inner


# When we call outer(), it returns the inner function.
f = outer()

# Even though outer() has finished executing, f() still remembers 'x'!
print(f())
```

### 2. Creating a State (The Counter)
Because closures remember their enclosing variables, we can use them to maintain state without needing to write a full Object-Oriented Class.

> **Note:** We use the `nonlocal` keyword to tell Python that we want to modify the variable that lives in the outer function, rather than creating a new local variable.

```python
def make_counter():
    count = 0  # enclosed variable

    def counter():
        nonlocal count  # Declares intent to modify the outer 'count' variable
        count += 1
        return count

    return counter


# Create an instance of our counter
my_counter = make_counter()

print("First call: ", my_counter())  # 1
print("Second call:", my_counter())  # 2
```

### 3. Inspecting the Backpack
Where is this state actually stored? Python stores these captured variables (called "free variables") in a special hidden attribute called `__closure__`.

```python
# Inspecting the 'my_counter' closure
print("Closure tuple:", my_counter.__closure__)

# Let's peek inside the cell contents!
print("Current value inside the closure:", my_counter.__closure__[0].cell_contents)
```

### 4. Function Factories
Closures are perfect for "Function Factories" - functions that build and return other specialized functions.

```python
def make_multiplier(factor):
    # enclosing scope variable: factor
    def multiplier(number):
        return number * factor  # inner function references 'factor'

    return multiplier  # return the inner function


# We generate two entirely new, customized functions!
double = make_multiplier(2)
triple = make_multiplier(3)

print("Double of 5:", double(5))  # 10
print("Triple of 5:", triple(5))  # 15
```
