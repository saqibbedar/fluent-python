# Conditional Expressions (Ternary Operator)

Python official documentation calls this a `Conditional Expression`.
Developers also frequently refer to it as an **inline if-else statement** or a **ternary operator/expression**.

It allows you to test a condition in a single line replacing the multiline if-else, making your code highly compact and readable.

### Syntax
```python
variable = value_if_true if condition else value_if_false
```

````python
is_raining = True

# Traditional multi-line way
if is_raining:
    action = "Stay inside"
else:
    action = "Go outside"

print("Traditional:", action)

# The Pythonic Ternary way
action = "Stay inside" if is_raining else "Go outside"
print("Ternary:", action)
````

### Direct Execution
You don't always have to assign it to a variable. You can execute functions directly.

````python
score = 49

# Print directly based on a condition
print("Pass") if score >= 50 else print("Fail")
````

### Using the `or` Operator for Defaults
A very common pattern in Python is setting a default value if a variable is `None`, empty, or evaluates to `False`. The `or` operator is often used as a shorthand ternary for this.

````python
user_role = None  # Imagine this comes from a database, but could be empty

# The 'or' operator evaluates the left side. If truthy, it returns it.
# If falsy (like None, 0, or ""), it returns the right side.
username = user_role or "Guest"
print("Role (using or):", username)

# Strict checking (if you ONLY want to fall back if it is explicitly None)
username_strict = user_role if user_role is not None else "Guest"
print("Role (strict):  ", username_strict)
````
