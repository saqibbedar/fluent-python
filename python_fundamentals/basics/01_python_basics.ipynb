## 1. Printing & Variables

The `print()` function is your best friend. It displays output on the screen.
Variables are like named boxes where you can store data.

```python
print("Hello, world!")

# Let's create some variables
name = "Saqib Bedar"  # A string (text)
age = 22              # An integer (whole number)
height = 5.4          # A float (decimal number)
is_student = True     # A boolean (True or False)

print(f"Hello, my name is {name} and I am {age} years old.")
```

> **Tip:** Notice the `f` before the string in the print statement? That's an **f-string**. It allows you to inject variables directly into text using curly braces `{}`. It's the modern, preferred way to format strings in Python.

## 2. Data Types & User Input

You can always check the type of a variable using the `type()` function.

When you use `input()` to get information from a user, Python *always* treats it as a string. If you need a number, you have to **cast** (convert) it.

```python
print(type(name))  # <class 'str'>
print(type(age))   # <class 'int'>

# Run this cell and type your age when prompted!
user_age_str = input("What's your age? ")

# If we try to add 1 to user_age_str, Python would crash because it's text.
# We must cast it to an integer first:
user_age = int(user_age_str)

print(f"Wow, you will be {user_age + 1} next year!")
```

## 3. Operators

Python has several types of operators. Let's explore the most common ones.

```python
a = 10
b = 3

print("--- Arithmetic ---")
print("Addition:      ", a + b)   # 13
print("Division:      ", a / b)   # 3.3333... (always returns a float)
print("Floor Division:", a // b)  # 3 (removes the decimal)
print("Modulus:       ", a % b)   # 1 (the remainder)
print("Exponent:      ", a ** b)  # 1000 (10 to the power of 3)

print("\n--- Comparison ---")
print("Is a equal to b? ", a == b)  # False
print("Is a not eq to b?", a != b)  # True
print("Is a > b?        ", a > b)   # True
```

### Logical, Identity, and Membership Operators
These are crucial for writing readable conditions.

```python
# Logical Operators (and, or, not)
has_id = True
print("Logical AND:", age >= 18 and has_id) 

# Identity Operators (is, is not) - checks if they are the exact same object in memory
x = None
print("Identity:", x is None)

# Membership Operators (in, not in) - checks if a value is inside a sequence
fruits = ["apple", "banana", "mango"]
print("Membership:", "mango" in fruits)
```

## 4. Conditionals (`if`, `elif`, `else`)

Conditionals allow your code to make decisions based on the state of your variables.

```python
score = 75

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

### A Practical Example: ATM Withdrawal
Let's put it all together. Variables, casting, conditionals, and assignment operators!

```python
balance = 50000.00
withdraw = float(input("Amount to withdraw: "))

if balance <= 0:
    print("Account frozen: Insufficient Balance")
elif withdraw > balance:
    print("Transaction failed: Insufficient funds")
else:
    if withdraw < 500.0:
        print("Withdrawal amount is too small. Minimum is 500.")
    else:
        balance -= withdraw  # This is the same as: balance = balance - withdraw
        print(f"Withdrawn! Remaining balance: {balance:.2f}")
```
