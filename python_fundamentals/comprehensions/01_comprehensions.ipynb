# Comprehensions in Python

Comprehensions provide a concise and readable way to create collections (like lists, dictionaries, and sets) out of other iterables. They are one of the most Pythonic features of the language.

Instead of writing a multi-line `for` loop with `append()`, you can do it all in a single elegant line.

---

## 1. List Comprehensions

### Basic Syntax
```python
new_list = [expression for item in iterable]
```

````python
numbers = [1, 2, 3, 4, 5]

# Traditional Loop
squares_loop = []
for n in numbers:
    squares_loop.append(n**2)

# List Comprehension
squares_comp = [n**2 for n in numbers]

print("Loop:         ", squares_loop)
print("Comprehension:", squares_comp)
````

### Filtering with `if` (Conditionals)

You can filter items by adding an `if` condition at the end.

```python
new_list = [expression for item in iterable if condition]
```

````python
# Keep only the even numbers
evens = [n for n in numbers if n % 2 == 0]
print("Evens:", evens)
````

### Transforming with `if-else` (Ternary)

If you want to apply an `if-else` condition to *transform* the output (not just filter), the `if-else` goes *before* the `for` loop.

```python
new_list = [value_if_true if condition else value_if_false for item in iterable]
```

````python
# Label numbers as 'Even' or 'Odd'
labels = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print("Labels:", labels)
````

### Nested Loops (Matrix Flattening)
You can chain multiple `for` loops inside a comprehension. Read them from left to right.

````python
matrix = [[1, 2], [3, 4], [5, 6]]

# Read as: "Give me 'num' for each 'row' in 'matrix', and for each 'num' in 'row'"
flat = [num for row in matrix for num in row]
print("Flattened:", flat)
````

---

## 2. Dictionary Comprehensions

You can build dictionaries dynamically using `{key: value ...}` syntax.

````python
names = ["Alice", "Bob", "Charlie"]

# Create a dict mapping names to their string lengths
name_lengths = {name: len(name) for name in names}
print("Dict Comprehension:", name_lengths)
````

---

## 3. Set Comprehensions

Similar to dictionary comprehensions but without the colons. It uses curly braces `{}` and automatically removes duplicates.

````python
duplicates = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

# Create a set of the squares of unique numbers
unique_squares = {n**2 for n in duplicates}
print("Set Comprehension:", unique_squares)
````

---

## 4. Generator Expressions

If you use parentheses `()` instead of brackets `[]`, you create a **Generator Expression** instead of a List Comprehension.

**Why use this?** A list comprehension generates the *entire list* in memory at once. A generator expression computes values *one by one* on the fly, which is incredibly memory efficient for massive datasets.

````python
import sys

# Creates all 10,000 items in memory instantly
list_comp = [x**2 for x in range(10000)]

# Creates a generator object that produces items only when asked
gen_expr = (x**2 for x in range(10000))

print(f"List Comp memory: {sys.getsizeof(list_comp)} bytes")
print(f"Gen Expr memory:  {sys.getsizeof(gen_expr)} bytes")

# You can iterate over the generator just like a list
print("First value from generator:", next(gen_expr))
````
