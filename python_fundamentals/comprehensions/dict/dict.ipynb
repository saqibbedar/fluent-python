# dict comprehension

Syntax:

```py
new_dict = {key_expression: value_expression for item in iterable}
```

---

## basic dict comprehension

````python
# Map names to their text lengths

names = ["Alice", "Bob", "Charlie"]

# Traditional way
name_lengths = {}
for name in names:
    name_lengths[name] = len(name)

print(f"traditional way: {name_lengths}")

# comprehension way
name_lengths = {name: len(name) for name in names}
print(f"comprehension way: {name_lengths}")
````

## Merging Two Lists into a Dictionary (Using zip)

````python
keys = ["id", "role", "status"]
values = [101, "admin", "active"]

# comprehension way
user_profile = {k: v for k, v in zip(keys, values)}
print(user_profile)
````

# Conditional

````python
prices = {"apple": 5, "banana": 12, "mango": 20}

# Keep if price > 10, then apply 10% discount
discounted_price = {item: price * 0.9 for item, price in prices.items() if price > 10}

print(discounted_price)
````

> Note: We use .items() to loop through both the keys and values of the original dictionary at the same time.
