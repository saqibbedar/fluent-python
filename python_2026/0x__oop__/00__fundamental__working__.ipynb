# OOP in Python

Object oriented programming (OOP) in Python is a programming paradigm that uses objects to represent data and methods.

### Class and Object

- **Class**: A blueprint or template for creating other objects.
- **Object**: An instance of a class containing real data.



### Example

```python
class Person:
    # constructor
    def __init__(self, name, age, height):
        # instance variables
        self.name = name
        self.age = age
        self.height = height

    # generator
    def get(self):
        yield self.name
        yield self.age
        yield self.height

    # setter
    def set(self, name, age, height):
        if self.name != name:
            self.name = name
        if self.age != age:
            self.age = age
        if self.height != height:
            self.height = height


# Creating an object or instance of class Person
p1 = Person("Saqib Bedar", 22, 5.4)

print(f"{p1.__dict__}\n")


# generator
for val in p1.get():
    print(val)
```

### Verify `p1` and `p2`

```python
# setter
p2 = p1                 # p2 has now the same object and both p1 and p2 are pointing at it.

print(f"{p1=}\n{p2=}\n")
print(f"Is same Address? {p1 is p2}\nAre both equal? {p1 == p2}")

import ctypes

count1 = ctypes.c_long.from_address(id(p1)).value
count2 = ctypes.c_long.from_address(id(p2)).value
count3 = ctypes.c_long.from_address(id(Person)).value

print(f"obj_refcnt_p1: {count1}")
print(f"obj_refcnt_p2: {count2}")
print(f"obj_refcnt_p2: {count3}")
```

### Adding arbitrary attributes and changing name attribute 

Let's modify `p2`, we are going to add first rename attribute name and then we gonna add new arbitrary attribute on `p2` and verify both `__ dict __` to look if change is global that means change is occurring in both or `p2` only. If change is happening in both then they are pointing at one object with singular `__ dict __` and `p1 & p2` are just labels. 

```python
p2.name = "xyz"
p2.city = "Larkana"
print(f"\n{p2.__dict__}")
print(f"\n{p1.__dict__}")
```

```python
# formatting heading
print(f"{"Index":<10} | {"Key":<10} | {"Value":<10}")
print("-"*34)

for i, (k, v) in enumerate(p1.__dict__.items()):
    print(f"{i:<10} | {k:<10} | {v:<10}")
```

# Dataclass

```python
import random
from dataclasses import dataclass

@dataclass
class Student:
    # constructor
    id: int
    name: str
    dept: str

    # generator
    def get(self):
        yield self.id
        yield self.name
        yield self.dept

    # setter
    def update(self, id=None, name=None, dept=None):
        self.id = id if id is not None else self.id
        self.name = name if name is not None else self.name
        self.dept = dept if dept is not None else self.dept

std = [Student(i, f"STD-{i}", random.choice(["AI", "CS", "DS", "SE", "IT"])) for i in range(1, 6)]

print(std[0])
```

```python
# unpack generator
id, name, dept = std[0].get()

print(f"Student 1: {id, name, dept}")
```

```python
# Unpack all students

for s in std:
    id, name, dept = s.get()
    print(f"ID: {id}, Name: {name}, Dept: {dept}")
```

```python
# duplicates checks
def has_duplicate_ids(std_lst):
    # generator expression
    ids = (std.id for std in std_lst)

    # A set removes duplicates; if sizes match, all elements are unique
    return len(std_lst) != len(set(ids))

# check duplicate id
if has_duplicate_ids(std):
    print("Warning: Duplicate student IDs found!")
else:
    print("Success: All student IDs are unique.")
```
