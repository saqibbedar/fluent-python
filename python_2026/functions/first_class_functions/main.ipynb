# First-Class Functions

In Python, **functions are first-class objects (or first-class citizens)**, meaning **they are treated just like any other data type, such as strings, integers, or lists**. This means you can assign them to variables, pass them as arguments to other functions, return them from other functions, and store them in data structures.

---

# Core Properties of First-Class Functions

### 1. Assigning Functions to Variables

You can assign a function reference to a variable without executing it (by omitting the parentheses). Both the original function name and the new variable will point to the same memory object.

```python
def greet(name):
    return f"Hello, {name}"

# Assign the function reference to a variable
say_hello = greet

# Call the function using the variable
print(say_hello("Saqib"))
```

### 2. Passing Functions as Arguments

Because functions are objects, you can feed them into other functions. A function that accepts another function as an argument is known as a **higher-order function**.

```python
def greeting(name):
    return f"Hello, {name}"

# Higher order function that accepts a function as 'func_argument'
def execute_greeting(func_argument, username):
    return func_argument(username)

# Pass the 'greeting' function object into 'execute_greeting'
print(execute_greeting(greeting, "Saqib Bedar"))
```

### 3. Returning Functions from Functions

A function can dynamically generate and return another inner function. This is heavily utilized to build function factories and closures.

```python
# closure
def multiplier_factory(factor):
    # Inner function that captures 'factor' variable
    def multiply_by(number): 
        return number * factor
    return multiply_by  # Return the inner function without calling it

double = multiplier_factory(2)
triple = multiplier_factory(3)

print(double(10))
print(triple(10))
```

### 4. Storing Functions in Data Structures

Functions can live inside collections like lists, dictionaries, or tuples. You can iterate through them or look them up programmatically.

```python
def add(a, b): return a + b
def sub(a, b): return a - b
def mul(a, b): return a * b

# store functions as values inside a dictionary (remember JS)
math_operators = {
    "add": add,
    "sub": sub,
    "mul": mul
}

# Access and call the function 
result = math_operators["add"](2, 2)

print(result)
```
