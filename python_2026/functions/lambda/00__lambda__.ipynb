# Lambda Function in Python

If you are aware of `JS` arrow function then Python's lambada function is actually the same with few tweaks. It is a anonymous function that is defined without a name and contains only a single expression. Used for short-lived, throwaway logic. 

#### Syntax
```md
lambda arguments: expression
```

# Quick comparison between regular and lambda function

````python
# Regular function expression
def power(x):
    return x ** 2           # equivalent to pow(x, 2) or import math and use math.pow(x, 2) and it will return float number

# lambda function
power_lambda = lambda x: x ** 2

print(f"{power(5)=}")
print(f"{power_lambda(5)=}")
````

# Common use cases

It excels as temporary arguments passed into higher-order functions like `map()`, `filter()`, and `sorted()` etc.

### 1. Using with `map(function, iterable)`

````python
# list
numbers_lst = [1, 2, 3, 4]

squared_lst = list(map(lambda x: x ** 2, numbers_lst))

print(f"{squared_lst=}")
````

````python
numbers_tuple = (1, 2, 3, 4)

squared_tuple = tuple(map(lambda x: x ** 2, numbers_tuple))

print(f"{squared_tuple=}")
````

````python
numbers_set = {1, 2, 3, 4}

squared_set = set(map(lambda x: x ** 2, numbers_set))

print(f"{squared_set=}")
````

### 2. Using with `filter(function, iterable)`

````python
numbers_lst = [1, 2, 3, 4, 5, 6]

even_nums_lst = list(filter(lambda x: x % 2 == 0, numbers_lst))

print(f"{even_nums_lst=}")
````

````python
numbers_tup = [1, 2, 3, 4, 5, 6]

even_nums_tup = tuple(filter(lambda x: x % 2 == 0, numbers_tup))

print(f"{even_nums_tup=}")
````

````python
numbers_set = {1, 2, 3, 4, 5, 6}

even_nums_set = set(filter(lambda x: x % 2 == 0, numbers_set))

print(f"{even_nums_set=}")
````

````python
# prices are in PKR
prices = {
    "apple": 350,
    "banana": 150,
    "cherry": 450
}

expensive_items = dict(filter(lambda item: item[1] > 200, prices.items()))

print(f"{expensive_items=}")
````
