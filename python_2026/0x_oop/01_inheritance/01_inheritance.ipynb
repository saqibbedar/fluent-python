# Basic Syntax

To inherit from another class, pass the parent class name inside parentheses when defining the child class.

### Example

```py
# Parent Class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    
    def move(self):
        return "Moving forward"

# Child class inherits from Parent class Vehicle
class Car(Vehicle):
    def honk(self):
        return "Beep beep!"

my_car = Car("Toyota")

print(my_car.move())            # Output: Moving forward (Inherited)
print(my_car.honk())            # Output: Beep beep! (Child specific)
```

# The `super()` Function

The `super()` function allows a child class to call methods from its parent class. This is most frequently used within the `__ init __()` constructor to cleanly initialize parent attributes without duplicating initialization logic.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Developer(Employee):
    def __init__(self, name, salary, programming_lang):
        # Initializes name and salary using the Employee parent constructor
        super().__init__(name, salary) 
        self.programming_lang = programming_lang
````

# Method Overriding

If a child class defines a method with the exact same name as a method in its parent class, the child's implementation overrides the parent's. Python will prioritize executing the child's version of that method.

````python
class Animal:
    def make_sound(self):
        return "Generic animal sound"

class Dog(Animal):
    def make_sound(self):
        return "Woof!"      # Overrides the parent method
````

# Perfect Class Inheritance Example

````python
class Student:
    def __init__(self, id, name, department):
        self.id = id
        self.name = name
        self.department = department

    # generator
    def get_std(self):
        yield self.id
        yield self.name
        yield self.department

    # Even better as or operator is ambiguous with falsy values
    def set_std(self, id=None, name=None, department=None):
        self.id = id if id is not None else self.id
        self.name = name if name is not None else self.name
        self.department = department if department is not None else self.department
````

````python
class GraduateStudent(Student):
    # first initialize all parent attributes
    # Student attributes: id, name, department
    # this attributes: thesis, faculty_advisor, research_area
    def __init__(self, id, name, department, thesis, faculty_advisor, research_area):
        # 1. Initialize parent attributes using super()
        super().__init__(id, name, department)

        # 2. Initialize child-specific attributes
        self.thesis = thesis
        self.faculty_advisor = faculty_advisor
        self.research_area = research_area

    # 1. Method Overriding: Extending the generator for GraduateStudent
    def get_std(self):
        # Yield parent attributes first
        yield from super().get_std()

        # yield child attributes
        yield self.thesis
        yield self.faculty_advisor
        yield self.research_area

    # 2. Method Overriding for the setter
    def set_std(self, id=None, name=None, department=None, thesis=None, faculty_advisor=None, research_area=None):
        # 1. Call parent setter to handle id, name, and department
        super().set_std(id, name, department)

        # 2. Handle child-specific attributes using same logic (if-else statement expression)
        self.thesis = thesis if thesis is not None else self.thesis
        self.faculty_advisor = faculty_advisor if faculty_advisor is not None else self.faculty_advisor
        self.research_area = research_area if research_area is not None else self.research_area

# Universally Unique Identifiers
import uuid         # for unique id
 
# Inherited class
grd_std = GraduateStudent(id=str(uuid.uuid4()), name="John Doe", department="CS", thesis="Neural Networks", faculty_advisor="Dr. Smith", research_area="Deep Learning")

# Use the inherited setter method
grd_std.set_std(name="Alice Vance")

# unpack generator
id, name, dept, thesis, fac_adv, rsch_area = grd_std.get_std()

print(f"Graduate Student: {id, name, thesis, fac_adv, rsch_area}")
````

# Types of Inheritance

Python is highly versatile and supports five structural patterns of inheritance:

- **Single**: A single child class inherits from one parent class.

- **Multiple**: A child class inherits directly from more than one parent class.

- **Multilevel**: A child class inherits from a parent, which itself is a child of another class.

- **Hierarchical**: Multiple unique child classes all inherit from the same single parent class.

- **Hybrid**: A combined mixture of two or more of the inheritance types listed above.

---

# Method Resolution Order (MRO)

In multiple inheritance scenarios, if different parent classes share identical method names, Python determines which one to execute using the **Method Resolution Order (MRO)**. Python resolves this left-to-right, depth-first, ensuring it does not visit any parent class twice. You can inspect this path by calling `ClassName.mro()` or using the `__ mro __` attribute.

# Helpful Built-in Functions

- **`isinstance(obj, Class)`**: Returns True if an object is an instance of that class or its subclasses.

- **`issubclass(Child, Parent)`**: Returns True if the first class is a direct or indirect subclass of the second.

````python
print(GraduateStudent.mro())        # return a list
print(GraduateStudent.__mro__)      # return a tuple
````

````python
class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return "Hello from B"

class C(A):
    def greet(self):
        return "Hello from C"

class D(B, C):
    pass

obj = D()
print(obj.greet())  # What will this print?
print(D.mro())
````
