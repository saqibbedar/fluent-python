# `or` Operator

In Python, the or operator does not look for None. It looks for truthiness.

**The Problem with Falsy Values**

In Python, all of the following values are considered falsy:

- None

- 0 (the integer zero)

- "" (an empty string)

- [] (an empty list)

```python
class Student:
    def __init__(self, id, name, dep):
        self.id = id
        self.name = name
        self.dep = dep
    
    # The following code with or is dangerous, we will see below how it's gonna cause issues while argument passing
    def update(self, id=None, name=None, dep=None):
        self.id = id or self.id
        self.name = name or self.name
        self.dep = dep or self.dep

p1 = Student(5, "John Doe", "Computer Science")

print(p1.__dict__)
```

### Update `p1` to detect defects in `or` operators

```python
# We want to genuinely change p1 values
p1.update(0, "", "CS")
```

### Verify `p1`

```python
print(p1.__dict__)
```

> We updated the values but because of `0` and `""` are false in Python so L.H.S remains false and old value was used instead, even though we have given new values.

---

# Solution

The solution is simple and standard approach, use inline-if-else statements.

```python
class Student:
    def __init__(self, id, name, dep):
        self.id = id
        self.name = name
        self.dep = dep
    
    # The following code with or is dangerous, we will see below how it's gonna cause issues while argument passing
    def update(self, id=None, name=None, dep=None):
        self.id = id if id is not None else self.id
        self.name = name if name is not None else self.name
        self.dep = dep if dep is not None else dep

p1 = Student(5, "John Doe", "CS")
print(p1.__dict__)
```

### Modify values

```python
p1.update(id=0, name="", dep="CS")
```

### Verify

```python
print(p1.__dict__)
```

> Hence, falsy values assigned successfully, so a good practice is to use inline-if-else statement as short hands instead of using or in such scenarios, to strictly assign a new value.
