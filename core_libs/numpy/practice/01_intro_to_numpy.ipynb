# NumPy (Numerical Python)

```python
import numpy as np

# verify version
print(np.__version__)
```

---

### Create NumPy Array

```python
# create first numpy array
array = np.array([1, 2, 3, 4, 5])

print(array)
```

### Verify type/dimension/shape

```python
# Type/dimension/shape

# using python way
print(f"Type: {type(array)} || {type(array).__name__}")         # returns class name      

# standard way
print(f"Dimensions: {array.ndim}")                              # returns int
# get the shape
print(f"Shape: {array.shape}")                                  # returns tuple
```

---

### Multiplication

```python
# Note: Python list copies references or duplicates elements but NumPy does actual element-wise multiplication.

# multiply each element by 2
print(f"Before Multiplication: {array}")
array *= 2
print(f"After multiplying each element/item by 2: {array}")
```
