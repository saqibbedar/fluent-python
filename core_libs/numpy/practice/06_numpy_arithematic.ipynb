```python
import numpy as np
```

```python
# Scalar arithmetic

array = np.array([1, 2, 3])

print(array + 1)
print(array - 1)
print(array / 2)
print(array * 3)
print(array // 2)
print(array ** 2)
```

### Vectorized Math Functions

```python
arr = np.array([1, 2, 3, 4])

print(np.sqrt(arr))
```

```python
rng = np.random.default_rng()

arr = np.full(shape=10, fill_value=2.3)

# Generates 10 distinct random numbers and multiplies element-by-element
arr = rng.random(size=arr.size)

print(arr)
```

```python
print(np.round(arr))
```

```python
# to round down
print(np.floor(arr))
```

```python
# to round up
print(np.ceil(arr))
```
