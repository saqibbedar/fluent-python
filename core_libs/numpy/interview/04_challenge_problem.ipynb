Imagine you are managing an online store. You have two datasets:

   1. sales Matrix: How many units of Item X and Item Y were sold on Monday and Tuesday.
   2. prices Matrix: The Base Price and the Tax Amount for Item X and Item Y.

    [2,  3]   # Tuesday: 2 of Item X, 3 of Item Y

# 2. Prices: Rows = [Item X, Item Y], Columns = [Base Price, Tax]

---

# Solution

````python
import numpy as np

sales = np.array([[5, 10], [2, 3]])
prices = np.array([[10, 2], [8, 1]])

# STEP A: Use true matrix multiplication (@)
daily_totals = sales @ prices  

# STEP B: Sum rows to get total revenue (Base + Tax) per day
total_revenue_per_day = daily_totals.sum(axis=1) 
print("Correct Total Revenue per day (Mon, Tue):", total_revenue_per_day)
# Output: [150  51] (Monday made $150 total, Tuesday made $51)

# STEP C: Apply the mask to total_revenue_per_day
days_over_100 = total_revenue_per_day[total_revenue_per_day > 100]

print("Revenues over 100:", days_over_100)
# Output: [150]
````
