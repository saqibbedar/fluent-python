> Generators are one-and-done: once used, they don't exist anymore

# Generator Expressions (The "Tuple" Comprehension)

You might think that wrapping a comprehension in parentheses `()` creates a tuple comprehension. However, Python uses that syntax to create a **Generator Expression**.

Instead of building a full collection in memory all at once, a generator creates items **one by one, on demand (lazily)**. This makes them incredibly memory-efficient when dealing with massive datasets.

---

# 1. The Syntax and Difference

If you print a list comprehension, you see the data. If you print a generator expression, you see the generator object itself.

```python
# List Comprehension (Immediately allocates memory for 1 million items)
list_comp = [x**2 for x in range(1000000)]

# Generator Expression (Allocates zero data memory upfront)
gen_exp = (x**2 for x in range(1000000))

print(list_comp[:3])  # Output: [0, 1, 4]
print(gen_exp)  # Output: <generator object <genexpr> at 0x...>
```

---

# 2. How to Get Values Out of a Generator

Because generators are lazy, you must explicitly ask for the data using a loop, the next() function, or by passing it to a collection constructor.

```python
numbers = range(1, 4)
squares_gen = (n**2 for n in numbers)

# Method A: Using next()
print(next(squares_gen))  # Output: 1
print(next(squares_gen))  # Output: 4

# Method B: Looping through the rest
for val in squares_gen:
    print(val)  # Output: 9
```

---

# 3. Turning a Generator into a Tuple

```python
# Creating an actual tuple
evens_tuple = tuple(x for x in range(10) if x % 2 == 0)

print(evens_tuple)
```

## Iterating

```python
numbers_gen = (x**2 for x in range(1, 4))  # Generates 1, 4, 9

# Standard for loop way
for num in numbers_gen:
    print(num)
```

```python
# next() way
colors_gen = (c for c in ["red", "green", "blue"])

print(next(colors_gen))  # Output: red
print(next(colors_gen))  # Output: green

# You can still loop through whatever is left!
for remaining in colors_gen:
    print(remaining)  # Output: blue
```
