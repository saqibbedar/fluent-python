# List Comprehensions (Basic Syntax)

Syntax: 
```py
new_list = [expression for item in iterable]
```

---

## 1. Traditional Loop vs. List Comprehension

````python
numbers = [1, 2, 3, 4, 5]
squares = []

for n in numbers:
    squares.append(n**2)

print(squares)  # Output: [1, 4, 9, 16, 25]
````

## List comprehension

````python
numbers = [1, 2, 3, 4, 5]
squares = [n**2 for n in numbers]

print(squares)
````

---

## 2. List Comprehensions with Conditionals

Syntax:

```py
new_list = [expression for item in iterable if condition]
```

````python
# Keep even numbers only

# Traditional way
evens = []

for n in numbers:
    if n % 2 == 0:
        evens.append(n)

print(f"traditional way: {evens}")

# Comprehension way
c_evens = [n for n in numbers if n % 2 == 0]
print(f"comprehension way: {c_evens}")
````

> How to read it: Loop through numbers, check if n % 2 == 0, and if true, keep n.

## The if-else (Transforming)

Syntax
```py
new_list = [expression_if_true if condition else expression_if_false for item in iterable]
```

````python
numbers = [1, 2, 3, 4]

# Traditional way
labels = []

for n in numbers:
    if n % 2 == 0:
        labels.append("Even")
    else:
        labels.append("Odd")

print(f"traditional way: {labels}")

# Comprehension Way
c_labels = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print(f"comprehension way: {c_labels}")
````

> Evaluate "Even" if n % 2 == 0 else "Odd" for every n inside the loop. \
> Note: it's just if-else comprehension first and then looping

# 3. Matrix / Nested List Comprehensions

````python
matrix = [[4, 5, 6]]

# Traditional Way
flat = []
for row in matrix:
    for num in row:
        flat.append(num)


print(f"traditional way: {flat}")


# Comprehension way
flat = [num for row in matrix for num in row]

print(f"traditional way: {flat}")
````

> How to read it: Read the loops from left to right, exactly like a sentence: "Give me num for each row in matrix for each num in that row." \
> OR, loop through matrix and get the row and loop through row
