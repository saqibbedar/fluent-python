# set comprehensions

Syntax:

```py
new_set = {expression for item in iterable}
```

---

## 1. Basic Set Comprehension (Removing Duplicates)

````python
words = ["Apple", "banana", "APPLE", "Banana", "cherry"]

# Traditional Way
unique_words = set()
for w in words:
    unique_words.add(w.lower())

print(f"traditional way: {unique_words}")

# Comprehension way
unique_words = {w.lower() for w in words}
print(f"comprehension way: {unique_words}")
````

---

## 2. Set Comprehension with Filtering (if)

````python
names = ["Al", "Alice", "Bob", "Charlie", "Amanda"]

# extract first letter if length > 3
initials = {name[0] for name in names if len(name) > 3}

print(initials)
````
