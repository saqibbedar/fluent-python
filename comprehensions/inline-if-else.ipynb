# Naming convention for `Conditional Expression`

Python official documentation says; `Conditional Expression`.

Developers also frequently refer to it as an `inline if-else statement` or a `ternary expression`.


### Syntax

```py
variable = value_if_true if condition else value_if_false
```

<br>

---

### Quick Example

````python
# normal way of writing conditions
is_raining = True

if is_raining:
    print("Stay in the room")
else:
    print("Go outside")
````

````python
# in-case you want to save the output (which is most common)
output = "Stay in the room" if is_raining else "Go outside"
print(output)

# Or print directly
print("Stay in the room") if is_raining else print("Go outside")
````

---

# Examples

**1. Variable Assignment**

````python
age = 20
status = "Adult" if age >= 20 else "Minor"
print(status)
````

**2. Inside a print function**

````python
score = 49
print("Pass") if score >= 50 else print("Fail")
````

---

# Using with `or` operator

**1. Using `or`**

````python
user_role = None        # can be admin etc
username  = user_role or "Guest"            # if leftside true assign it else assign default value 
print(username)
````

**2. Strict None checking (Safe if `0` or `""` are valid entries)**

````python
username = user_role if user_role is not None else "Guest"
print(username)
````

---

# Using with `list`

````python
labels = ["Even" if x % 2 == 0 else "Odd" for x in range(5)]

print(labels)
````
