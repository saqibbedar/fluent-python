# List

Syntax

```py
new_list = [<item> for <item> in <iterable> if <condition>]
```

List Comprehension (With Transformation/Ternary)

```py
new_list = [<value_if_true> if <condition> else <value_if_false> for <item> in <iterable>]
```

---

# Dict

```py
# Syntax for filtering elements
new_dict = {key: value for key, value in <iterable> if <condition>}

# Syntax for changing values inline
new_dict = {key: (<val_if_true> if <condition> else <val_if_false>) for key, value in <iterable>}
```
