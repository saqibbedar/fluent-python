# Challenge 1: The Conditional Expression (Ternary)

Convert this standard if-else block into a single line of code.

```py
score = 85

if score >= 70:
    result = "Pass"
else:
    result = "Fail"
```

````python
score = 85

result = "Pass" if score >= 70 else "Fail"

print(result)
````

# Challenge 2: List Comprehension (Filtering)

Convert this `for` loop that filters items into a single-line list comprehension.

```py
prices = [10, 45, 120, 80, 200, 30]
expensive_prices = []

for p in prices:
    if p > 50:
        expensive_prices.append(p)
```

````python
prices = [10, 45, 120, 80, 200, 30]
expensive_prices = [p for p in prices if p > 50]
print(expensive_prices)
````

# Challenge 3: List Comprehension (Transformation)

Convert this loop into a one-liner. Hint: This one requires combining the loop with a conditional expression at the front!

```py
numbers = [1, 2, 3, 4, 5]
categories = []

for n in numbers:
    if n % 2 == 0:
        categories.append("Even")
    else:
        categories.append("Odd")
```

````python
numbers = [1, 2, 3, 4, 5]
categories = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print(categories)
````
